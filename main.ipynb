```python
series_metadata = pd.read_csv("data/series_metadata.csv")
series_metadata.head()
```

```python
metadata_dict = (
    series_metadata
    .set_index("series_name")
    [["genres", "origin_country", "keywords"]]
    .to_dict("index")
)
```

```python
def get_category(series_name):
    stats = metadata_dict.get(series_name)

    if stats is None:
        return "other"

    genres = str(stats["genres"])
    origin = str(stats["origin_country"])
    keywords = str(stats["keywords"]).lower()

    if "Animation" in genres:
        if "JP" in origin:
            return "anime"
        elif "CN" in origin:
            return "donghua"
        elif "US" in origin:
            return "american_animation"
        elif "GB" in origin:
            return "british_animation"
        elif "AU" in origin:
            return "australian_animation"
        elif "FR" in origin:
            return "french_animation"
        else:
            return "other_animation"
    elif "Sci-Fi & Fantasy" in genres:
        return "fantasy_show"
    elif "Drama" in genres:
        if "US" in origin:
            return "american_drama"
        elif "GB" in origin:
            return "british_drama"
        elif "CN" in origin:
            return "chinese_drama"
        elif "IN" in origin:
            return "indian_drama"
        elif "TR" in origin:
            return "turkish_drama"
        else:
            return "other_drama"
    elif "Kids" in genres:
        return "kids_show"
    elif "Talk" in genres:
        return "talk_show"
    elif "Reality" in genres:
        return "reality_show"
    elif "Documentary" in genres:
        return "documentary"
    elif "miniseries" in keywords:
        return "miniseries"
    elif "sitcom" in keywords:
        return "sitcom"
    else:
        return "other_live_action"

top_1000["category"] = top_1000["series_name"].map(get_category)
```

```python
def clean_genres(genres):
    if pd.isna(genres):
        return []

    # If genres is already a list
    if isinstance(genres, list):
        return genres

    # If stored as a string representation of a list
    if isinstance(genres, str):
        try:
            import ast
            parsed = ast.literal_eval(genres)

            if isinstance(parsed, list):
                return parsed

            return [parsed]
        except (ValueError, SyntaxError):
            # Handle comma-separated strings
            return [g.strip() for g in genres.split(",")]

    return []
```
